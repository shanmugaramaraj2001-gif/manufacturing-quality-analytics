```python
import pandas as pd
print(pd.__version__)
```

```python
df = pd.read_csv("inspection_data.csv")
```

```python
df.head()

df.info()

df.shape
```

```python
df.info()

df.isnull().sum()

df.duplicated().sum()
```

```python
df.isnull().sum()
```

```python
df[df["Defect_Type"].isnull()]
```

```python
df["Inspection_Date"] = pd.to_datetime(
    df["Inspection_Date"]
)

df.info()
```

```python
df['Inspection_Date'] = pd.to_datetime(
    df['Inspection_Date'],
    format='%d-%m-%Y'
)
```

```python
df['Inspection_Date'].head()
```

```python
df["Defect_Type"] = df["Defect_Type"].fillna(
    "No Defect"
)
```

```python
df.isnull().sum()
```

Rejection Rate

```python
df["Rejection_Rate"] = (
    df["Qty_Rejected"] /
    df["Qty_Inspected"]
) * 100
```

Month Name

```python
df["Month"] = (
    df["Inspection_Date"]
      .dt.strftime("%b")
)
```

Month Number

```python
df["Month_No"] = (
    df["Inspection_Date"]
      .dt.month
)
```

Year

```python
df["Year"] = (
    df["Inspection_Date"]
      .dt.year
)
```

```python
df.head()
```

```python
df["Rejection_Rate"] = df["Rejection_Rate"].round(2)
```

```python
df.head()
```

Total Quantity inspected, Total Quantity Rejected & Total COPQ

```python
print("Total Inspected :", df["Qty_Inspected"].sum())

print("Total Rejected :", df["Qty_Rejected"].sum())

print("Total COPQ :", round(df["Rejection_Cost"].sum(), 2))
```

```python
overall_rejection_rate = (
    df["Qty_Rejected"].sum()
    /
    df["Qty_Inspected"].sum()
) * 100

round(overall_rejection_rate, 2)
```

```python
df.to_csv(
    "inspection_data_cleaned.csv",
    index=False
)
```

```python
import pymysql
import sqlalchemy
```

```python
import os
from sqlalchemy import create_engine
from sqlalchemy.engine import URL

# Set these environment variables locally before running this cell.
# Never commit database credentials to GitHub.
url = URL.create(
    drivername="mysql+pymysql",
    username=os.environ.get("MYSQL_USER", "root"),
    password=os.environ["MYSQL_PASSWORD"],
    host=os.environ.get("MYSQL_HOST", "localhost"),
    port=int(os.environ.get("MYSQL_PORT", "3306")),
    database=os.environ.get("MYSQL_DATABASE", "aerospace_quality")
)

engine = create_engine(url)
```

```python
engine
```

```python
df.to_sql(
    "inspection_data",
    con=engine,
    if_exists="replace",
    index=False
)
```
